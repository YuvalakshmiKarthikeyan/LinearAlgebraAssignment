```python
import numpy as np
vector_a = np.array([2, 4, 6])
vector_b = np.array([1, 3, 5])
matrix_a = np.array([
[2, 1, 3],
[1, 0, 2],
[4, 1, 1]
])
matrix_b = np.array([
[1, 2, 0],
[3, 1, 4],
[2, 5, 1]
])
```

```python
# Using NumPy:
# ● Perform vector addition, subtraction, and dot product using vector_a and vector_b
#Addition
Vec_Addition = vector_a + vector_b
print("Vector Addition :",Vec_Addition)
#Subtraction
Vec_Subtraction = vector_a - vector_b
print("Vector Subtraction :",Vec_Subtraction)
Vec_dot=np.dot(vector_a,vector_b)
print("Vector Dot Product :",Vec_dot)
```

```python
# ● Perform matrix addition and matrix multiplication using matrix_a and matrix_b
print("Matrix Addition :" ,matrix_a + matrix_b)
```

```python
Mat_Mul=np.matmul(matrix_a,matrix_b)
print("Matrix Multiplication :",Mat_Mul)
```

```python
# ● Calculate determinant of matrix_a
# ● If determinant of matrix_a is not equal to zero, then find inverse of matrix_a
det = np.linalg.det(matrix_a)
print("Determinant:", det)
if det != 0:
    print("Inverse:\n", np.linalg.inv(matrix_a))
```

```python
# ● Compute eigenvalues and eigenvectors of matrix_a
eigenvalues,eigenvectors = np.linalg.eig(matrix_a)
print("EigenValues:",eigenvalues)
print("EigenVectors:",eigenvectors)
```

```python
# # Calculate probability of successful sales
# Demonstrate conditional probability using customer segments
# Probability of sales success and failure
sales_success = np.array([0.65, 0.35])

# Customer segments
customer_segments = ["Premium", "Regular"]

# Conditional probabilities P(Purchase | Segment)
segment_purchase_probability = {
    "Premium": 0.80,
    "Regular": 0.50
}

# Overall probability of successful sales
prob_success = sales_success[0]

print("Probability of Successful Sales:", prob_success)

# Conditional probabilities
for segment in customer_segments:
    print(f"P(Purchase | {segment}) = {segment_purchase_probability[segment]:.2f}")
```

```python
# Formula: P(A|B) = [P(B|A) * P(A)] / P(B)
# P(A): Probability that a customer is Premium
P_A = 0.40
# P(B): Probability that a customer makes a purchase
P_B = 0.62
# P(B|A): Probability that a Premium customer makes a purchase
P_B_given_A = 0.80

P_A_given_B = (P_B_given_A * P_A) / P_B
print("probability of a customer purchase : ",P_A_given_B)
```

```python
# Statistical Dataset
sales_data = [120, 135, 150, 145, 160, 155, 170, 165, 180, 175,
190, 200, 210, 195, 185, 175, 165, 155, 145, 135]
```

```python
#Using NumPy / Pandas / statistics library:
# ● Compute mean, median, and mode of sales_data
import pandas as pd
import statistics as stats
import matplotlib.pyplot as plt
import math
mean = np.mean(sales_data)
print("Mean:", mean)
```

```python
median = np.median(sales_data)
print("Median:", median)
```

```python
mode = stats.mode(sales_data)
print("Mode:", mode)
```

```python
# ● Calculate range, variance, and standard deviation
range = np.max(sales_data) - np.min(sales_data)
print("Range:", range)
```

```python
variance = np.var(sales_data)
print("Variance:", variance)
```

```python
std_dev = np.std(sales_data)
print("Standard Deviation:", std_dev)
```

```python
# To Calculate skewness
# skewness = mean - mode/Standard Deviation
skewness =  (mean - mode)/std_dev
print("Skewness:", skewness)
# Result > 0: Positive Skew (Right Skew)
# Result < 0: Negative Skew (Left Skew)
# Result = 0: No Skew (Perfectly Symmetrical
if(skewness > 0):
  print("Graph is Postively Skewed(Right Skew)")
elif(skewness < 0):
  print("Graph is Negatively Skewed(Left Skew)")
else:
  print("Graph is Symmetrical")
```

```python
# Measure skewness and kurtosis
from scipy.stats import kurtosis
kurtosis_value = kurtosis(sales_data)
print("Kurtosis:", kurtosis_value)
# Kurtosis = 0 (Mesokurtic)
# Kurtosis > 0 (Leptokurtic)
# Kurtosis < 0 (Platykurtic)
if(kurtosis_value > 0):
  print("Graph is Leptokurtic")
elif(kurtosis_value < 0):
  print("Graph is Platykurtic")
else:
  print("Graph is Mesokurtic")
```

```python
plt.hist(kurtosis_value, bins=6)
plt.xlabel("kurtosis")
plt.ylabel("Frequency")
plt.title("Histogram of Sales Data")
plt.show()
```
